```python
# VERSION 1 
note1_ue1 = float(input("Note 1 UE1 : "))
coeff1_ue1 = float(input("Coeff 1 UE1 : "))
note2_ue1 = float(input("Note 2 UE1 : "))
coeff2_ue1 = float(input("Coeff 2 UE1 : "))

moyenne_ue1 = (note1_ue1 * coeff1_ue1 + note2_ue1 * coeff2_ue1) / (coeff1_ue1 + coeff2_ue1)
print("Moyenne UE1 :", moyenne_ue1)
```
